# Data cleaning Notebook 

Suivant une nouvelle méthodologie, ce notebook orchestre simplement le travail. Toute la logique de nettoyage se trouve dans le fichier réutilisable `src/data_cleaning.py et data_loader.py` !

```python
#Importation
import sys
from pathlib import Path

src_path = Path.cwd().parent / "src"
sys.path.append(str(src_path))

from utils import setup_logger, ensure_directory_exists
from data_loading import load_data
from data_cleaning import clean_missing_values, clean_total_charges, convert_churn_to_binary, add_derived_columns


#confi du looger
logger = setup_logger("../reports/logs/data_cleaning.log")
```

- Chargement des données

```python
# Chemins des fichiers
INPUT_PATH = str(Path.cwd().parent / "data/raw/WA_Fn_Use_C_Telco_Customer_Churn.csv")
OUTPUT_PATH = str(Path.cwd().parent / "data/processed/churn_cleaned.csv")

#loading of data
df = load_data(INPUT_PATH)

#Appercu
print("Appercu des données: \n")
df.head()
```

```python
print(df.info())
```

```python
print(df.describe())
```

```python
print("\n===Missing values===")
print(df.isnull().sum())
```

```python
print("\nDuplicate rows:", df.duplicated().sum())
```

- Nettoyage des données

```python
# Nettoyage des données
df = clean_missing_values(df)
df = clean_total_charges(df)
df = convert_churn_to_binary(df)
df = add_derived_columns(df)

# Aperçu après nettoyage
print("Aperçu après nettoyage :")
display(df.head(10))
```

> Apres nettoyage et conversion en binaire; on vois aussi bien que la colunne `tenure_group` fut ajouté avec succes rendandant la taille `5 rows × 22 columns`

- Sauvegarde des données nettoyées

```python
# Sauvegarde des données nettoyées
ensure_directory_exists(str(Path(OUTPUT_PATH).parent))
df.to_csv(OUTPUT_PATH, index=False)
logger.info(f"Data saved in {OUTPUT_PATH}")
```
